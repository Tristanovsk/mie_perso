```python
#!/usr/bin/python
# -*- coding: utf-8 -*-

from __future__ import unicode_literals

import os
import glob
import pandas as pd
#import seaborn as sns

import yaml
from parse import *
import sys
import cmath

import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

rc = {"font.family" : "serif",'font.weight': 'normal',
      "mathtext.fontset" : "stix"}
plt.rcParams.update(rc)
plt.rcParams["font.serif"] = ["Times New Roman"] + plt.rcParams["font.serif"]
plt.rcParams.update({
                     'font.size': 16, 'axes.labelsize': 18,
                     })
import refractive_index.refractiveindex as ri
import refractive_index.water_refractive_index.auxdata as water

u = ri.utils()
wri = water.water_refractive_index()

opj =os.path.join
```

```python
def NtoEps(matVec):
    matVec = np.array(matVec)
    lamb = matVec[:, 0]
    n = matVec[:, 1]
    k = matVec[:, 2]

    N = n[:] + 1j * k[:]
    eps = N[:] * N[:]

    matEps = [lamb, eps.real, eps.imag]
    return matEps

figdir = '/DATA/projet/garaba/OP3/work/refractive_index/fig'
dirdata = ri.__path__[0]
subdir = 'database/data/organic/*)n -*/'
files = glob.glob(opj(dirdata,subdir,'*.yml'))
range = (0.350, 2.495)
wls = np.linspace(range[0], range[1], 1000)
water_temp,water_sal = 10,30
water_n =wri.n_harmel(wls*1000,water_temp,water_sal)

plt.figure()
plt.plot(wls,water_n)
plt.show()


#------------------------------
# Load refractive index database
# for desired files
dfs = []
for i,file in enumerate(files):
    material = file.split('/')[-2]
    print(material)
    basename = os.path.basename(file)
    print(i,basename)
    try:
        refrac = u.getData(file, wls)#/water_n
        if np.isnan(refrac).sum() > 200:
            continue
        df=pd.DataFrame({'wavelength':wls,'m':refrac})
        df['material']=material
        df['name']=basename
        dfs.append(df)
    except:
        continue

df_refrac = pd.concat(dfs)
df_refrac['n']=df_refrac.m.values.real
df_refrac['k']=df_refrac.m.values.imag

df_mean = df_refrac.groupby(by=['material','wavelength']).mean()
```

```python
fig,axs = plt.subplots(nrows=2,ncols=2,figsize=(12,12))
axs=axs.ravel()
for id,g in df_mean.groupby('material'):
    g.reset_index(inplace=True)
    axs[0].plot(g.wavelength,g.n,label=id)
    axs[1].plot(g.wavelength,g.k,label=id)
    g =g[g.wavelength<0.9]
    axs[2].plot(g.wavelength,g.n,label=id)
    axs[3].plot(g.wavelength,g.k,label=id)
axs[1].legend()

plt.show()

fig,axs = plt.subplots(nrows=2,ncols=3,figsize=(15,7))

axs=axs.ravel()

for id,g in df_mean.groupby('material'):
    g.reset_index(inplace=True)
    g['water_n']=water_n
    axs[0].plot(g.wavelength,g.n,label=id)
    axs[1].plot(g.wavelength, g.n / g['water_n'], label=id)
    axs[2].plot(g.wavelength,g.k,label=id)
    g =g[g.wavelength<0.9]
    axs[3].plot(g.wavelength,g.n,label=id)
    axs[4].plot(g.wavelength,g.n/g.water_n,label=id)
    axs[5].plot(g.wavelength,g.k,label=id)
axs[2].legend(fontsize=10)
for ax in axs:
    ax.set_xlabel('Wavelength (µm)')
axs[0].set_ylabel('real part (vacuum)')
axs[1].set_ylabel('real part (rel. to water)')
axs[2].set_ylabel('imaginary part')
axs[3].set_ylabel('real part (vacuum)')
axs[4].set_ylabel('real part (rel. to water)')
axs[5].set_ylabel('imaginary part')
fig.suptitle('Refractive index of some common polymers')
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
#plt.savefig(opj(figdir,'polymer_refrac_index.png'),dpi=300)
plt.show()
```
